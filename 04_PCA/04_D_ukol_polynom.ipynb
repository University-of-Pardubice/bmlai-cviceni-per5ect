# Načtení dat

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
data = pd.read_csv ("../dataset/prices.csv",  sep=';')
```

# Lineární model
- standardizujte data
- vytvořte a natrénujte lineární model
- zobrazte parametry modelu
- vizualizujte lineární model
- zobrazte skóre modelu

```python
data.head()
```

```python
# - standardizujte data
X = data[['Size']] 
y = data['Price']   

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# - vytvořte a natrénujte lineární model
model = LinearRegression()
model.fit(X_scaled, y)
```

```python
y_pred = model.predict(X_scaled)
```

```python
# - vizualizujte lineární model
plt.figure(figsize=(8,6))
sns.scatterplot(x=X_scaled.flatten(), y=y)
plt.plot(X_scaled, y_pred, color='red', linewidth=2)
plt.xlabel("Size")
plt.ylabel("Price")
plt.legend()
plt.show()
```

```python
# - zobrazte skóre modelu
r2 = r2_score(y, y_pred)
print(f"R²: {r2:.4f}")
```
